## MLFlow Tracking Server

```python
import mlflow
```

```python
# Set the tracking URI to the MLflow server
# Whatever experiments and runs we create will be logged to this server
mlflow.set_tracking_uri("http://127.0.0.1:5000")
```

```python
# Set the experiment name. An experiment is a collection of runs, and is the primary way to organize runs in MLflow.
mlflow.set_experiment("Check localhost connection")

# Start a new MLflow run.
with mlflow.start_run():
    mlflow.log_metric("test", 1)
    mlflow.log_metric("Mudit", 2)
```

```python
# Start another MLflow run.
with mlflow.start_run():
    mlflow.log_metric("test1", 3)
    mlflow.log_metric("Mudit1", 4)
```
